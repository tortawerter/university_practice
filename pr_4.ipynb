```python
#8 
import string

alp = string.digits + string.ascii_uppercase


def is_valid(number, system):
    global alp

    if system < 2 or system > 36:
        return False, "основание вне диапазона 2..36"

    for char in number:
        if char not in alp:
            return False, char
        if alp.index(char) >= system:
            return False, char

    return True, ""


number = input("Введите число: ").upper()
system = int(input("Введите основание системы(до 36 ричной): "))

ok, issue = is_valid(number, system)

if ok:
    print(f"Число '{number}' корректно для системы с основанием {system}")
else:
    if issue in alp:
        print(f"Ошибка: цифра '{issue}' недопустима для системы с основанием {system}")
    else:
        print(f"Ошибка: {issue}")
```

```python
#9
import string
import math

alp = string.digits + string.ascii_uppercase


def from_ten(number, system):
    global alp

    if number == 0:
        return "0"
    else:
        digits = []
        n = number
        while n > 0:
            digits.append(str(n % system))
            n //= system
        result = ''
        for i in digits:
            result += alp[int(i)]

    return result[::-1]


def to_ten(number, system):
    global alp

    ans = 0
    for i, digit in enumerate(number[::-1]):
        digit = alp.index(digit)
        ans += int(digit) * (system ** i)

    return ans


amount = 4
nums = []
for i in range(amount):
    nums.append(10**(i + 1))
dif = 0
for i in nums:
    print(f"\nЧисло: {i}")
    print(f"{'Осн.':>5} | {'Запись':<20} | {'L эксп':>6} | {'L теор':>6} | {'Разница':>4}")
    print("-" * 60)

    for system in range(2, 37):
        test = from_ten(i, system)
        teory = math.floor(math.log(i, system)) + 1
        if abs(len(test) - teory) != 0:
            dif += 1
        print(f"{system:<5} | {test:<20} | {len(test):<6} | {teory:<6} | {abs(len(test) - teory):<4}")

print("Количество расхождений: ", dif)
# ближайшую степень основания к числу   
```

```python
#10
import string

alp = string.digits + string.ascii_uppercase


def from_ten_div(number, system):
    global alp

    operations = 0

    if number == 0:
        return "0", operations
    else:
        digits = []
        n = number
        while n > 0:
            digits.append(str(n % system))
            n //= system
            operations += 1
        result = ''
        for i in digits:
            result += alp[int(i)]
            operations += 1

    return result[::-1], operations


def from_ten_pow(number, system):
    global alp

    operations = 0

    if number == 0:
        return "0", operations

    top_power = 0
    while system ** (top_power + 1) <= number:
        top_power += 1
        operations += 1

    result = ''
    n = number
    for power in range(top_power, -1, -1):
        weight = system ** power
        digit = n // weight
        n -= digit * weight
        result += alp[digit]
        operations += 1

    return result, operations


nums = [10, 100, 1000, 10000]
systems = [2, 8, 10, 16, 36]

print(f"{'Число':>9} | {'Осн.':>4} | {'Деление':<14} | {'Степени':<14} | {'Совпад.':>6} | {'Опер. дел':>9} | {'Опер. сте':>9}")
print("-" * 90)

for number in nums:
    for system in systems:
        word_div, ops_div = from_ten_div(number, system)
        word_pow, ops_pow = from_ten_pow(number, system)

        same = "да" if word_div == word_pow else "НЕТ"

        print(f"{number:>9} | {system:>4} | {word_div:<14} | {word_pow:<14} | {same:>6} | {ops_div:>9} | {ops_pow:>9}")


print("\nСуммарные операции по всем тестам:")

total_div = 0
total_pow = 0

for number in nums:
    for system in systems:
        _, ops_div = from_ten_div(number, system)
        _, ops_pow = from_ten_pow(number, system)
        total_div += ops_div
        total_pow += ops_pow

print(f"Метод деления:  {total_div}")
print(f"Метод степеней: {total_pow}")
```
