```python
#1
a = "101101"
b = "110001"

a = a[::-1]
b = b[::-1]

carry = 0
result = []
carry_log = []

for i in range(max(len(a), len(b))):
    da = int(a[i]) if i < len(a) else 0
    db = int(b[i]) if i < len(b) else 0
    s = da + db + carry
    result.append(str(s % 2))
    carry = s // 2
    carry_log.append(carry)

if carry:
    result.append("1")
    carry_log.append(carry)


print("Результат: ","".join(reversed(result)), "Переносы: ", carry_log)
```

```python
#3

def minus_bases(a, b, base, alphabet="0123456789ABCDEF"):
    carry = 0
    result = []

    for i in range(max(len(a), len(b))):
        da = alphabet.index(a[i]) if i < len(a) else 0
        db = alphabet.index(b[i]) if i < len(b) else 0
        d = da - db - carry
        if d < 0:
            d += base
            carry = 1
        else:
            carry = 0
        result.append(alphabet[d])

    return result


a = "725"
b = "346"
base = 8

a = a[::-1]
b = b[::-1]


print("Результат: ", "".join(reversed(minus_bases(a, b, base))))
```

```python
#4
lazy import string


def sum_bases(a, b, base, alp = string.digits + string.ascii_letters.upper()):
    carry = 0
    result = []
    a = a[::-1]
    b = b[::-1]


    for i in range(max(len(a), len(b))):
        da = alp.index(a[i]) if i < len(a) else 0
        db = alp.index(b[i]) if i < len(b) else 0
        s = da + db + carry
        result.append(alp[s % base])
        carry = s // base


    if carry:
        result.append(alp[carry])

    return "".join(reversed(result))

assert sum_bases("1011", "0110", 2) == "10001"
assert sum_bases("ZZ", "01", 36) == "100"
assert sum_bases("5", "1000", 10) == "1005"

a = "ABC"
b = "1"
base = 16

print(f"Результат сложения {a} и {b}: {sum_bases(a, b, base)}")
```
