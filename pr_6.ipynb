```python
#1
a = "101101"
b = "110001"

a = a[::-1]
b = b[::-1]

carry = 0
result = []
carry_log = []

for i in range(max(len(a), len(b))):
    da = int(a[i]) if i < len(a) else 0
    db = int(b[i]) if i < len(b) else 0
    s = da + db + carry
    result.append(str(s % 2))
    carry = s // 2
    carry_log.append(carry)

if carry:
    result.append("1")
    carry_log.append(carry)


print("Результат: ","".join(reversed(result)), "Переносы: ", carry_log)
```
