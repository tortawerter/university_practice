```python
#7
lazy import string


def sum_bases(a, b, base, alp=string.digits + string.ascii_uppercase):
    carry = 0
    result = []
    carries = []
    a = a[::-1]
    b = b[::-1]

    for i in range(max(len(a), len(b))):
        da = alp.index(a[i]) if i < len(a) else 0
        db = alp.index(b[i]) if i < len(b) else 0
        s = da + db + carry
        result.append(alp[s % base])

        carry = s // base
        if carry:
            carries.append(i)

    if carry:
        result.append(alp[carry])
        carries.append(max(len(a), len(b)))

    return "".join(reversed(result)), carries


def to_base(n, base, alp=string.digits + string.ascii_uppercase):
    result = ""

    while n > 0:
        result = alp[n % base] + result
        n //= base

    return result


a = "101111"
b = "1011010010"
x = int(a, 2)
y = int(b, 2)

for base in range(2, 37):
    a = to_base(x, base)
    b = to_base(y, base)

    result, carries = sum_bases(a, b, base)

    print(f"Основание {base}:")
    print(f"{a} + {b} = {result}")
    print(f"Переносов: {len(carries)}, позиции: {carries}")
    print()

    
```

```python
#10
lazy import string


def to_base(n, base, alp=string.digits + string.ascii_uppercase):
    result = ""
    while n > 0:
        result = alp[n % base] + result
        n //= base

    return result


def from_base(s, base, alp=string.digits + string.ascii_uppercase):
    result = 0
    for ch in s:
        result = result * base + alp.index(ch)
    return result



def sum_ok(a, b, base, alp=string.digits + string.ascii_uppercase):
    carry = 0
    result = []

    a = a[::-1]
    b = b[::-1]

    for i in range(max(len(a), len(b))):
        da = alp.index(a[i]) if i < len(a) else 0
        db = alp.index(b[i]) if i < len(b) else 0
        s = da + db + carry
        result.append(alp[s % base])
        carry = s // base

    if carry:
        result.append(alp[carry])

    return "".join(reversed(result))



def sum_broken(a, b, base, alp=string.digits + string.ascii_uppercase):
    carry = 0
    result = []

    a = a[::-1]
    b = b[::-1]

    for i in range(max(len(a), len(b))):
        da = alp.index(a[i]) if i < len(a) else 0
        db = alp.index(b[i]) if i < len(b) else 0
        s = da + db + carry
        result.append(alp[s % base])
        carry = s % base


    return "".join(reversed(result))


def check(a, b, base):
    ok = sum_ok(a, b, base)
    bad = sum_broken(a, b, base)

    v_ok = from_base(ok, base)
    v_bad = from_base(bad, base)
    v_true = from_base(a, base) + from_base(b, base)

    print(f"  {a} + {b} (base {base})")
    print(f"    эталон:    {ok}  = {v_ok}")
    print(f"    искажён:   {bad}  = {v_bad}")
    print(f"    ожидалось:       {v_true}")

    if v_ok != v_true:
        print("    ОШИБКА В ЭТАЛОНЕ!")
    if v_bad != v_true:
        print("    обнаружено расхождение в искажённом алгоритме")
    else:
        print("    искажение не проявилось")
    print()


print("=== Одиночные проверки ===")
check("1011", "0110", 2)
check("777", "001", 8)
check("FF", "01", 16)
check("725", "346", 8)
```
