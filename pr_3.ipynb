```python
#9
import random


def code_hamming(word: str) -> str:
    p0 = int(word[0])
    p1 = int(word[1])
    p2 = int(word[2])
    p3 = int(word[3])
    
    hamming_word = ''.join(map(str, [p0 ^ p1 ^ p3, p0 ^ p2 ^ p3, p0, p1 ^ p2 ^ p3, p1, p2, p3]))

    return hamming_word


def mistake(word) -> str:
    word = list(word)
    idx = random.randint(0, len(word) - 1)  
    word[idx] = '1' if word[idx] == '0' else '0'
    return ''.join(word)


def decode_hammind(word, check: str) -> list:
    word = list(map(int, list(word)))
    s1 = word[0] ^ word[2] ^ word[4] ^ word[6]
    s2 = word[1] ^ word[2] ^ word[5] ^ word[6]
    s3 = word[3] ^ word[4] ^ word[5] ^ word[6]

    syndrom = int(''.join(map(str, (s1, s2, s3)[::-1])), 2)

    if syndrom == 0:
        return [0, 0]

    bug = word[syndrom - 1]

    if bug == 0:
        word[syndrom - 1] = 1
        if (str(word[2]) + str(word[4]) + str(word[5]) + str(word[6])) == check:  
            return [1, 1]
        else:
            return [1, 0]
    word[syndrom - 1] = 0
    if str(word[2]) + str(word[4]) + str(word[5]) + str(word[6]) == check:
        return [1, 1]
    else:
        return [1, 0]


stats = {"Бит четности: ": [0, 0],
         "Троирование: ": [0, 0],
         "Хэмминг: ": [0, 0]}


tests = 1000
for check in range(tests):
    decode = []

    word = ""
    for i in range(4):
        word += str(random.randint(0, 1))

    chetn_word = word + str(word.count("1") % 2)
    troir_word = ''.join([change * 3 for change in [i for i in word]])
    hamming_word = code_hamming(word)


    chetn_word = mistake(chetn_word)
    troir_word = mistake(troir_word)
    hamming_word = mistake(hamming_word)

    if chetn_word.count("1") % 2 != 0:
        stats["Бит четности: "][0] += 1

    for i in range(0, len(troir_word) - 2, 3):
        decode.append(troir_word[i:i + 3])

    flag = False

    for i in decode:
        if i != "111" or i != "000":
            flag = True

    for i in range(len(decode)):
        if decode[i].count("1") > 1:
            decode[i] = "1"
        else:
            decode[i] = "0"

    if flag:
        stats["Троирование: "][0] += 1

    if ''.join(decode) == word:
        stats["Троирование: "][1] += 1

    
    for pos, i in enumerate(decode_hammind(hamming_word, word)):
        stats["Хэмминг: "][pos] += i



print("Колличество тестов: ", tests)
print()

for name, data in stats.items():
    print(f"{name}")
    print(f"  Обнаружено: {data[0]}/{tests}")
    print(f"  Исправлено: {data[1]}/{tests}")
    print()




```
