```python
#9


def binary_to_octal(binary):
    while len(binary) % 3 != 0:
        binary = "0" + binary

    result = ""
    for i in range(0, len(binary), 3):
        result += str(int(binary[i:i+3], 2))

    return result


def binary_to_hex(binary):
    alphabet = "0123456789ABCDEF"

    while len(binary) % 4 != 0:
        binary = "0" + binary

    result = ""
    for i in range(0, len(binary), 4):
        result += alphabet[int(binary[i:i+4], 2)]

    return result


def clean(word):
    n = 0
    while word[n] != "1":
        n += 1
    return word[n:]



numbers = [
    "1",
    "101",
    "1010",
    "101101",
    "10110101",
    "1011010100"
]



for num in numbers:
    octal = binary_to_octal(num)
    hex_number = binary_to_hex(num)

    znach = len(clean(num))
    neznach = len(num) - znach

    print(f"Двоичное: {num}", )
    print(f"Длина: {len(num)}")
    print(f"Значащих: {znach}")
    print(f"Ведущих нулей: {neznach}")
    print(f"Восьмеричное: {octal} (цифр: {len(octal)})")
    print(f"Шестнадцатеричное: {hex_number} (цифр: {len(hex_number)})")


    oct_groups = []
    tmp = num
    while len(tmp) % 3 != 0:
        tmp = "0" + tmp
    for i in range(0, len(tmp), 3):
        oct_groups.append(tmp[i:i+3])

    hex_groups = []
    tmp = num
    while len(tmp) % 4 != 0:
        tmp = "0" + tmp
    for i in range(0, len(tmp), 4):
        hex_groups.append(tmp[i:i+4])

    print(f"Группировка по 3 эдемента {oct_groups}")
    print(f"Группировка по 4 элемента {hex_groups}")
    print()
```

```python
#10
def binary_to_octal(binary):
    while len(binary) % 3 != 0:
        binary = "0" + binary

    result = ""
    for i in range(0, len(binary), 3):
        result += str(int(binary[i:i+3], 2))

    return result


def binary_to_hex(binary):
    alphabet = "0123456789ABCDEF"

    while len(binary) % 4 != 0:
        binary = "0" + binary

    result = ""
    for i in range(0, len(binary), 4):
        result += alphabet[int(binary[i:i+4], 2)]

    return result


def octal_to_binary(octal):
    binary = ""
    for ch in octal:
        binary += format(int(ch), "03b")
    return binary


def hex_to_binary(hex_number):
    alphabet = "0123456789ABCDEF"
    binary = ""
    for ch in hex_number:
        binary += format(alphabet.index(ch), "04b")
    return binary


def clean(word):
    n = 0
    while word[n] != "1":
        n += 1
    return word[n:]



nums = [
    "1011010100",
    "00110101",
    "1000000000",
    "1111"
]


for word in nums:
    octal = binary_to_octal(word)
    hex_number = binary_to_hex(word)

    return_octal = octal_to_binary(octal)
    return_hex = hex_to_binary(hex_number)

    print(f"Исходное двоичное: {word}")
    print(f"Восьмеричное: {octal}, обратно: {return_octal}")
    print(f"Шестнадцатеричное: {hex_number}, обратно: {return_hex}")

    same_oct = clean(word) == clean(return_octal)
    same_hex = clean(word) == clean(return_hex)

    print(f"Совпадение(8): {same_oct}")
    print(f"Совпадение(16): {same_hex}")

    if not same_oct:
        print("  Потеря информации в восьмеричном представлении!")
    if not same_hex:
        print("  Потеря информации в шестнадцатеричном представлении!")

    print()

print("Исследование ведущих нулей:")
for i in range(1, 4):
    test = i * "0" + nums[0]
    octal = binary_to_octal(test)
    hex_number = binary_to_hex(test)
    print(f"{test} - 8: {octal} | 16: {hex_number}")
```
