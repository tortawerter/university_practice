```python
#1

def to_twos(x, bits):
    if x >= 0:
        return x
    return (1 << bits) + x


def from_twos(v, bits):
    if v < (1 << (bits - 1)):
        return v
    return v - (1 << bits)


a = 23
b = -9
bits = 8

a_twos = to_twos(a, bits)
b_twos = to_twos(b, bits)

mask = (1 << bits) - 1
result = (a_twos + b_twos) & mask

print(f"{a} в доп. коде:  {format(a_twos, f'0{bits}b')}")
print(f"{b} в доп. коде: {format(b_twos, f'0{bits}b')}")
print(f"Результат:      {format(result, f'0{bits}b')}")
print(f"Значение:       {from_twos(result, bits)}")
```

```python
#3

def to_twos(x, bits):
    if x >= 0:
        return x
    return (1 << bits) + x


def add_with_overflow(a, b, bits):
    mask = (1 << bits) - 1
    a_twos = to_twos(a, bits) & mask
    b_twos = to_twos(b, bits) & mask
    
    result = (a_twos + b_twos) & mask

    sign_a = (a_twos >> (bits - 1)) & 1
    sign_b = (b_twos >> (bits - 1)) & 1
    sign_r = (result >> (bits - 1)) & 1

    over = (sign_a == sign_b) and (sign_a != sign_r)

    return result, over


bits = 8
print(1 << (bits - 1))

for a, b in [(23, -9), (100, 50), (-100, -200), (127, 1), (-128, -1), (10, 20)]:

    result, over = add_with_overflow(a, b, bits)

    if result < (1 << (bits - 1)):
        value = result
    else:
        value = result - (1 << bits)

    if over:
        flag = "ПЕРЕПОЛНЕНИЕ"
    else:
        flag = "ок"

    print(f"{a} + {b} = {value}  [{flag}]")
```

```python
#5 - 6

def to_bin_float_normal(a: str):
    integer, fract = list(map(int, a.split(".")))
    fract = fract / 10 ** len(str(fract))
    integer = bin(integer)[2:]
    coef = 0
    ans_fract = ""

    max_bits = 20
    i = 0
    while fract != 0 and i < max_bits:
        fract *= 2
        ans_fract += str(fract)[0]
        fract = fract % 1
        i += 1


    full = integer + ans_fract
    first_one = full.index("1")


    tail = full[first_one + 1:]
    mantissa = "1." + (tail if tail else "0")

    coef = len(integer) - first_one - 1

    return f"{mantissa} x 2**({coef}), ", "Периодичная" if i == max_bits else "Обычный вид"


a = "0.25"
print(*to_bin_float_normal(a))
```

```python
#7

def check(a: str):
    integer, fract = list(map(int, a.split(".")))
    fract = fract / 10 ** len(str(fract))
    integer_bin = bin(integer)[2:]

    ans_fract = ""
    max_bits = 20
    i = 0
    while fract != 0 and i < max_bits:
        fract *= 2
        ans_fract += str(fract)[0]
        fract = fract % 1
        i += 1


    truncated = len(ans_fract) == max_bits and fract != 0

    print(f"Число: {a}")
    print(f"Двоичная запись: {integer_bin}.{ans_fract}")
    if truncated:
        print(f"Дробь бесконечная, обрезана до {max_bits} бит")
    else:
        print("Дробь конечная, представима точно")
    print()


tests = ["0.5", "0.25", "0.125", "0.1", "0.2", "0.3", "0.75", "0.625"]

for a in tests:
    check(a)
```

```python
#9

def to_twos(x, bits):
    if x >= 0:
        return x
    return (1 << bits) + x


def from_twos(v, bits):
    if v < (1 << (bits - 1)):
        return v
    return v - (1 << bits)


def sign_of(v, bits):
    return (v >> (bits - 1)) & 1


def add_signed(a, b, bits):
    mask = (1 << bits) - 1
    a_twos = to_twos(a, bits) & mask
    b_twos = to_twos(b, bits) & mask

    result = (a_twos + b_twos) & mask

    sign_a = sign_of(a_twos, bits)
    sign_b = sign_of(b_twos, bits)
    sign_r = sign_of(result, bits)

    overflow = (sign_a == sign_b) and (sign_a != sign_r)

    return from_twos(result, bits), overflow


def report(a, b, bits, op):
    if op == "+":
        value, overflow = add_signed(a, b, bits)
    else:
        value, overflow = add_signed(a, -b, bits)

    flag = "ПЕРЕПОЛНЕНИЕ" if overflow else "ок"
    print(f"{a:>5} {op} {b:>5} = {value:>5}  [{flag}] (bits={bits})")


print("=== Сложение ===")
for a, b in [(23, -9), (100, 50), (-100, -50), (127, -1), (-128, -1)]:
    report(a, b, 8, "+")

print()
print("=== Вычитание ===")
for a, b in [(23, 9), (100, -50), (-100, 50), (-128, 1), (127, 1)]:
    report(a, b, 8, "-")
```
