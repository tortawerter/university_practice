```python
#7-8
import string


def sum_bases(a, b, base, alp = string.digits + string.ascii_letters.upper()):
    carry = 0
    result = []
    a = a[::-1]
    b = b[::-1]


    for i in range(max(len(a), len(b))):
        da = alp.index(a[i]) if i < len(a) else 0
        db = alp.index(b[i]) if i < len(b) else 0
        s = da + db + carry
        result.append(alp[s % base])
        carry = s // base


    if carry:
        result.append(alp[carry])

    return "".join(reversed(result))


number = -1
positive = 256
bits = 9

positive_binary = format(abs(number), f"0{bits}b")
inverted = "".join("1" if b == "0" else "0" for b in positive_binary)
twos = sum_bases(inverted, "1", 2)

sum_bits = sum_bases(twos, format(positive, f"0{bits}b"), 2)
sum_bits = sum_bits[-bits:]
result = int(sum_bits, 2)

if result >= 2 ** (bits - 1):
    result = result - 2 ** bits

min_value = -(2 ** (bits - 1))
max_value = 2 ** (bits - 1) - 1

print("Отрицательное число:", number)
print("Положительное число:", positive)
print()
print(f"Дополнительный код {number}:", twos)
print(f"Двоичный код {positive}: ", format(positive, f"0{bits}b"))
print()
print("Сумма в двоичном виде: ", sum_bits)
print("Результат в десятичном:", result)
print("Ожидаемый результат:  ", number + positive)
print("Дополнительный код корректен:", result == number + positive)
print()
print(f"Разрядность: {bits} бит")
print(f"Диапазон представимых значений: от {min_value} до {max_value}")
```

```python
#10
import string


def sum_bases(a, b, base, alp = string.digits + string.ascii_letters.upper()):
    carry = 0
    result = []
    a = a[::-1]
    b = b[::-1]


    for i in range(max(len(a), len(b))):
        da = alp.index(a[i]) if i < len(a) else 0
        db = alp.index(b[i]) if i < len(b) else 0
        s = da + db + carry
        result.append(alp[s % base])
        carry = s // base


    if carry:
        result.append(alp[carry])

    return "".join(reversed(result))


numbers = [-14, 13, -128, 127, -255, 200]
bits_list = [4, 8, 9, 16]

for bits in bits_list:
    min_value = -(2 ** (bits - 1))
    max_value = 2 ** (bits - 1) - 1

    print("=" * 45)
    print(f"Разрядность: {bits} бит")
    print(f"Диапазон: от {min_value} до {max_value}")
    print("=" * 45)

    for number in numbers:
        if number < min_value or number > max_value:
            print(f"  {number:>6} вне диапазона")
            continue

        if number >= 0:
            direct = format(number, f"0{bits}b")
            reverse = direct
            twos = direct
        else:
            direct = "1" + format(abs(number), f"0{bits - 1}b")
            reverse = direct[0] + "".join("1" if b == "0" else "0" for b in direct[1:])
            twos = sum_bases(reverse, "1", 2)

        if twos[0] == "0":
            twos_back = int(twos, 2)
        else:
            twos_back = int(twos, 2) - 2 ** bits

        correct = twos_back == number
        print(f"  {number:>6} {twos} {twos_back:>6} {correct}")

    print()
```
