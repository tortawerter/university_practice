```python
#1
number = 25
bits = 8

binary = format(number, f"0{bits}b")

print("Число:", number)
print("Разрядность:", bits)
print("Двоичное представление:", binary)
```

```python
#2
number = -13
bits = 8

number_abs = format(abs(number), f"0{bits-1}b")
direct_code = "1" + number_abs

print("Число:", number)
print("Модуль:", number_abs)
print("Прямой код:", direct_code)
```

```python
#3

number = -13
bits = 8

number_abs = format(abs(number), f"0{bits-1}b")
direct_code = "1" + number_abs
inverse_code = "".join("1" if b == "0" else "0" for b in direct_code)

print("Число:", number)
print("Прямой код:", direct_code)
print("Обратный код:", inverse_code)
```

```python
#4

number = -13
bits = 8

positive_binary = format(abs(number), f"0{bits}b")
inverted = "".join("1" if b == "0" else "0" for b in positive_binary)
twos = format(int(inverted, 2) + 1, f"0{bits}b")

print("Число:", number)
print("Модуль в 8 битах:", positive_binary)
print("Инверсия:", inverted)
print("Дополнительный код:", twos)
```

```python
#5

def codes(number, bits):

    nums_abs = format(abs(number), f"0{bits-1}b")
    direct = "1" + nums_abs
    inverse = "".join("1" if b == "0" else "0" for b in direct)
    twos = format(int(inverse, 2) + 1, f"0{bits}b")
    
    return direct, inverse, twos


number = -13
bits = 8

direct, inverse, twos = codes(number, bits)

print("Число:", number)
print("Прямой код:         ", direct)
print("Обратный код:       ", inverse)
print("Дополнительный код: ", twos)
```
